```python
import pandas as pd
```

```python
x = pd.DataFrame(wine_quality.data.features)
y = pd.DataFrame(wine_quality.data.targets)
print(x.shape, y.shape)
```

```python
df = pd.concat([x, y], axis=1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 8))
ax = sns.countplot(data=df, x='quality')
for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.show()
```

```python
for column in x.columns:
    plt.figure(figsize=(10, 6))
    sns.barplot(data=df, x='quality', y=column)
    plt.title(f'{column} vs Wine Quality')
    plt.xlabel('Wine Quality')
    plt.ylabel(column)
    plt.show()
```

```python
correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(12, 8))

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')

plt.title('Correlation Matrix of Wine Quality Dataset')
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(data=df, x='quality', y='alcohol')

plt.title('Alcohol Content vs Wine Quality')
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol Content")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(data=df, x='quality', y='volatile_acidity', marker='o')

plt.title('Volatile Acidity vs Wine Quality')
plt.xlabel('Wine Quality')
plt.ylabel('Volatile Acidity')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(data=df, x='quality', y='chlorides', marker='o')

plt.title('Chlorides vs Wine Quality')
plt.xlabel('Wine Quality')
plt.ylabel('Chlorides')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(data=df, x='quality', y='density', marker='o')

plt.title('Density vs Wine Quality')
plt.xlabel("Wine Quality")
plt.ylabel("Density")
plt.show()
```

A positive correlation was observed with alcohol, while negative correlations were found with density, chlorides, and volatile acidity.

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='residual_sugar', y='density', hue='quality')

plt.title('Residual Sugar vs Density')
plt.xlabel('Residual Sugar')
plt.ylabel('Density')
plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.scatterplot(data=df, x='total_sulfur_dioxide', y='free_sulfur_dioxide', hue='quality')

plt.title('Total Sulfur Dioxide vs Free Sulfur Dioxide')
plt.xlabel('Total Sulfur Dioxide')
plt.ylabel('Free Sulfur Dioxide')
plt.show()
```

# Feature Engineering

```python
def quality_category(quality):
    if quality <= 4:
        return 'Low'
    elif quality <= 6:
        return 'Medium'
    else:
        return 'High'

df['quality_category'] = df['quality'].apply(quality_category)
```

```python
df['total_acidity'] = df['fixed_acidity'] + df['volatile_acidity']
```
