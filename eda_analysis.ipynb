```python
from ucimlrepo import fetch_ucirepo
```

```python
wine_quality = fetch_ucirepo(id=186)
```

```python
print(wine_quality.metadata)
```

```python
print(wine_quality.variables)
```

```python
import pandas as pd
```

```python
x = pd.DataFrame(wine_quality.data.features)
y = pd.DataFrame(wine_quality.data.targets)
print(x.shape, y.shape)
```

```python
df = pd.concat([x, y], axis=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
df.head(10)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 8))
ax = sns.countplot(data=df, x='quality')
for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.show()
```

```python
df_high_quality = df[df['quality'] == 9]
```

```python
df_high_quality
```

```python
df_low_quality = df[df['quality'] == 3]
```

```python
df_low_quality.head()
```

```python
correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(12, 8))

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')

plt.title('Correlation Matrix of Wine Quality Dataset')
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='quality', y='alcohol')

plt.title('Boxplot of Alcohol Content by Wine Quality')
plt.show()
```
